```python
from sklearn import datasets
df = datasets.load_breast_cancer()
```

```python
print("Features: ", df.feature_names)
print("Labels: ", df.target_names)
```

```python
df.data.shape
```

```python
print(df.target)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.data, df.target, test_size=0.3,random_state=109) 
```

```python
from sklearn import svm
clf = svm.SVC(kernel='linear')
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
```

```python
from sklearn.naive_bayes import GaussianNB
model_GBN = GaussianNB()
model_GBN.fit(X_train, y_train)
y_pred2 = model_GBN.predict(X_test)
```

```python
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred2))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt =  DecisionTreeClassifier()
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf =  RandomForestClassifier()
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred))
```

```python
import xgboost as xgb
clf=xgb.XGBClassifier()
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
from sklearn import metrics
print("Accuracy:",metrics.accuracy_score(y_test, y_pred))
```
